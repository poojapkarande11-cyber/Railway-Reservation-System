```python
# Task 1: append() and inser()

Subjects= ["English" , "Hindi" , "Marathi" , "SCience"]
print("Original list:" , Subjects)

# Add one more subject at the end of the list.

Subjects.append("Maths")
print("After append:", Subjects)

#Add a subject at the 2nd position (index 1) in the list.
Subjects.insert(1,"History")
print("After insert:" , Subjects)
```

```python
# Task 2: remove(), pop() and clear()

Numbers = [100, 200, 300, 400, 500, 600]
print("List:" , Numbers)
# Remove one number from the list by its value.
Numbers.remove(200)
print("after remove:" , Numbers)

#Remove the last number from the list using a method that also returns the removed value.
removed_value = Numbers.pop()
print("After pop:" , Numbers)

#  Print the removed value
print(removed_value)
```

```python
#Task 3: sort(), reverse() and count()

Marks = [65, 95, 55, 45, 95, 85]
print("list:" , Marks)

# Sort the list in ascending order.
Marks.sort()
print("After_sort:" , Marks)

#Reverse the sorted list.
Marks.reverse()
print("After_reverse:" , Marks)

# Count how many times the repeated number appears in the list.

Count_95 = Marks.count(95)
print("90_count:" , Count_95, "times")

```

```python
#Task 4: index() and copy(

colors = ["Red" , "Yellow" , "Pink" , "Purple" , "Black"]
print("list:" , colors)

# Find the index (position) of one colour in the list.
position = colors.index("Yellow")
print("Index of Yellow:", position)
      
      
## Make a copy of the list with a new name.

New_colors = colors.copy()
print("copied list:" , New_colors)

#Change one item in the copied list and check whether the original list also changed.
New_colors[0]= "white"

print("original_list:" , colors)
print("copied_new_list:" , New_colors)
```

```python
#Task 5: Changing Case

# Store your full name in a variable called name.

name= "pooja ambre"

# Print your name in all UPPERCASE letters.
print("UPPERCASE:",name.upper())
      
# Print your name in all lowercase letters.
print("lowercase:",name.lower())

# Print your name with only the first letter of each word capitalised.
print("Title_case:" , name.title())
```

```python
# Task 6- strip(), replace() and split()

#  Create a string with extra spaces
sentense = " I Love Python "
print("original sentense:", sentense)

#  Remove the extra spaces from the beginning and end of the string.
sentense = sentense.strip()
print("After_strip:", sentense)

# Replace the word &quot;love&quot; with &quot;enjoy&quot; in the sentence.

sentense = sentense.replace( "Love" , "enjoy")
print("after_replace:", sentense)

## Split the sentence into a list of individual words.
words = sentense.split()
print("after_split:", words)

```

```python
#Task 8 :- join() and startswith()/endswith()
# Create a list of 4 city names

cities = ["Mumbai", "Pune" ,"Nashik" , "Nagpur"]

# Join the list into a single string separated by commas
City_strings = ",".join(cities)
print("cities:",City_strings)

# Name from Task 5
name = "pooja ambre"

# Check whether the name starts with a vowel
print("Starts with a vowel:", name.lower().startswith(("a", "e", "i", "o", "u")))

# Check whether the name ends with the expected letter
print("ends with 'e':", name.lower().endswith("e"))
```
