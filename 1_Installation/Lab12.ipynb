```python
##Task 1: Create Function Using Default Arguments

def reservation_status (passanger_name , ticket_status ="Confirmed"):
    print("passanger Name :",passanger_name)
    print("Ticket_Status :", ticket_status)


reservation_status("Rahul sharma")
reservation_status("Amit verma", "waiting")

```

```python
##Task 2: Use *args (Variable Length Arguments)

#Create a function to handle multiple passenger names

def passanger_booking(*passangers):
    print("passanger List:")
    for name in passangers:
        print("passanger:", name)


# call the function

passanger_booking(
    "Rahul sharma",
    "priya Singh",
    "Amit Verma",
    "Neha Kapoor"
)
```

```python
#Task 3 **kwargs (keyword variable arguments)

##Create a function to store passenger ticket details dynamically

def ticket_details(**details):
   print("Ticket Information:")
   for key, value in details.items():
       print(key, ":", value)


```

```python
# call the function.
       
ticket_details(
   passenger_name="Rahul Sharma",
   train_name="Rajdhani Express",
   seat_number="B1-21",
   fare=2500,
   status="Confirmed")
```

```python
# Task 4: Create Recursive Seat Countdown Function

def seat_countdown(seats):
   if seats == 0:
       print("All Seats Booked")
       return
   print("Remaining Seats :", seats)
   seat_countdown(seats - 1)


```

```python
seat_countdown(5)
```

```python
# Task 5: Create Recursive Passenger Verification Function.

def verify_passenger(passenger_number):
   if passenger_number == 0:
       print("Verification Completed")
       return
   print("Verifying Passenger :", passenger_number)
   verify_passenger(passenger_number - 1)

```

```python
# call the function
verify_passenger(3)
```

```python
## Task 6: Build Final Advanced Booking Module.

# Create final booking function

def final_booking(
   train_name="Rajdhani Express",
   *seat_numbers
):
   print("===================================")
   print(" Railway Reservation ")
   print("===================================")
   print("Train Name :", train_name)
   for seat in seat_numbers:
       print("Allocated Seat :", seat)

```

```python
# call the function

final_booking(
   "Rajdhani Express",
   "B1-21",
   "B1-22",
   "B1-23")
```
